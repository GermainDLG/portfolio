```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import numpy as np
import matplotlib.pyplot as plt
from src.config import ECCENTRICITIES, DATA_DIR
```

```python
series_by_e = dict()
for eccentricity in ECCENTRICITIES:
    filename = f"orbit_e{eccentricity}_r0.npz"
    path = DATA_DIR / filename
    data = np.load(path)
    series_by_e[eccentricity] = data
```

```python
fig, ax = plt.subplots(figsize=(7, 7))
for (e, data) in series_by_e.items():
    ax.plot(data["x"], data["y"], label=f"e={e}")
ax.set_aspect("equal")
ax.plot(0, 0, marker="+", color="black")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("generate_orbits test (x vs. y)")
ax.legend()
```

```python
fig, ax = plt.subplots(figsize=(7, 7))
for (e, data) in series_by_e.items():
    ax.plot(data["t"], data["r"], label=f"e={e}")
ax.set_xlabel("t")
ax.set_ylabel("r")
ax.set_title("generate_orbits test (t vs. r)")
ax.legend()
```

```python
for (e, data) in series_by_e.items():
    expected_p = 1 - e
    expected_a = 1 + e
    actual_min = data["r"].min()
    actual_max = data["r"].max()
    print("---------")
    print(f"expected_p: {expected_p}")
    print(f"actual_min: {actual_min}")
    print(f"expected_a: {expected_a}")
    print(f"actual_max: {actual_max}")
```
